# 01 — Data Collection

Fetches F1 race data from the Ergast API for seasons 2011–2023 and caches locally.

**Run once.** Subsequent runs read from `data/raw/` cache — no API calls made.

```python
import subprocess
import sys
from pathlib import Path

result = subprocess.run(
    [sys.executable, str(Path('..') / 'data' / 'fetch_data.py')],
    capture_output=True, text=True
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

## Quick inspection of outputs

```python
import pandas as pd
from pathlib import Path

DATA = Path('..') / 'data'

for name in ['races', 'results', 'pit_stops', 'lap_times', 'drivers']:
    df = pd.read_csv(DATA / f'{name}.csv')
    print(f"{name}.csv  →  {df.shape[0]:,} rows × {df.shape[1]} cols")
```
